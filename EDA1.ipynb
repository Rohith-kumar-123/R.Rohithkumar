CHECKING MISSING VALUES


```python
df.isnull().sum()
```

```python
# Fill missing values in numeric columns

numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column].fillna(df[column].mean(), inplace=True)

# Fill missing values in categorical columns

categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    df[column].fillna(df[column].mode()[0], inplace=True)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicates Removed")
```

# Clean Column Names

```python
df.columns = df.columns.str.strip()

print(df.columns)
```

# Data Transformation

```python
# Data Transformation

df['income_thousands'] = df['income'] / 1000

print(df[['income', 'income_thousands']].head())
```

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 18, 35, 50, 100],
    labels=['Young', 'Adult', 'Middle-aged', 'Senior']
)

print(df[['age', 'age_group']].head())
```

```python
df.info()
```

# UNIVARIATE VISUALISATIONS

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df, x='age', bins=20, kde=True)

plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.kdeplot(data=df, x='income_thousands', fill=True)

plt.title('Income Distribution in Thousands')
plt.xlabel('Income (Thousands)')
plt.ylabel('Density')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='age_group')

plt.title('Distribution of Age Groups')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.countplot(x='gender', data=df)
plt.title('Gender Distribution')
plt.xticks([0,1], ['female', 'male'])
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

# BIVARIATE VISUALIZATIONS

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x='age', y='visits')

plt.title('Age vs Visits')
plt.xlabel('Age')
plt.ylabel('Visits')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='gender', y='income', data=df)
plt.title('Income by Gender')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x='age_group', y='income')

plt.title('Income Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Income')
plt.xticks(rotation=20)
plt.show()
```

# MULTIVARIATE VISUALISATIONS

```python
plt.figure(figsize=(9,5))

sns.barplot(
    data=df,
    x='program',
    y='visits',
    hue='gender',
    errorbar=None
)

plt.title('Average Visits by Program and Gender')
plt.xlabel('Program')
plt.ylabel('Average Visits')
plt.show()
```

```python
corr = df[['age', 'income_thousands', 'visits']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x='age_group',
    hue='gender'
)

plt.title('Age Group Distribution by Gender')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.xticks(rotation=20)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='program',
    y='age',
    hue='gender',
    data=df
)

plt.xticks([0,1], ['AFDC','SSI'])
plt.title('Age Distribution by Program and Gender')
plt.xlabel('Program')
plt.ylabel('Age')

plt.show()
```
