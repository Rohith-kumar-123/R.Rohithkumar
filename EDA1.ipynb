CHECKING MISSING VALUES


```python
df.isnull().sum()
```

```python
# Fill missing values in numeric columns

numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column].fillna(df[column].mean(), inplace=True)

# Fill missing values in categorical columns

categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    df[column].fillna(df[column].mode()[0], inplace=True)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicates Removed")
```

# Clean Column Names

```python
df.columns = df.columns.str.strip()

print(df.columns)
```

# Data Transformation

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

categorical_columns = ['married','gender','ethnicity','enroll','program']

for column in categorical_columns:
    df[column] = encoder.fit_transform(df[column])

df.head()
```

# UNIVARIATE VISUALISATIONS

```python
plt.figure(figsize=(6,4))
plt.hist(df['age'], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(df['income'])
plt.title("Income Boxplot")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.countplot(x='gender', data=df)
plt.title('Gender Distribution')
plt.xticks([0,1], ['female', 'male'])
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

# BIVARIATE VISUALIZATIONS

```python
plt.figure(figsize=(8,5))
sns.barplot(x='program', y='visits', data=df, estimator='mean', errorbar=None)
plt.xticks([0,1], ['AFDC', 'SSI'])
plt.title('Average Hospital Visits by Program')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='gender', y='income', data=df)
plt.title('Income by Gender')
plt.xticks([0,1], ['female', 'male'])
plt.show()
```

# MULTIVARIATE VISUALISATIONS

CORRELATION MATRIX

```python
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap='coolwarm',
            fmt='.2f')

plt.title('Correlation Heatmap')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x='program',
    y='visits',
    hue='gender',
    data=df,
    estimator='mean',
    errorbar=None
)

plt.xticks([0,1], ['AFDC','SSI'])
plt.legend(['Female','Male'])   # Adjust if needed

plt.title('Average Visits by Program and Gender')
plt.xlabel('Program')
plt.ylabel('Average Visits')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='program',
    y='age',
    hue='gender',
    data=df
)

plt.xticks([0,1], ['AFDC','SSI'])
plt.title('Age Distribution by Program and Gender')
plt.xlabel('Program')
plt.ylabel('Age')

plt.show()
```
