```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Features selected for clustering

cluster_features = [
    'age',
    'income',
    'visits',
    'children',
    'exposure',
    'health1',
    'health2',
    'access',
    'school'
]

X = df[cluster_features].copy()

# Remove missing values
X = X.dropna()

print("Shape of clustering data:", X.shape)
display(X.head())
```

```python
# Standardize the variables

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=cluster_features,
    index=X.index
)

display(X_scaled.head())
```

```python
# Elbow Method

inertia = []

K = range(2, 11)

for k in K:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(K, inertia, marker='o')

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal K")
plt.xticks(K)
plt.grid(True)

plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

silhouette_scores = []

K = range(2, 11)

for k in K:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)

# Find the best K
optimal_k = K[silhouette_scores.index(max(silhouette_scores))]

# Plot
plt.figure(figsize=(10, 6))

plt.plot(
    K,
    silhouette_scores,
    marker='o',
    linewidth=2
)

# Highlight best K
plt.axvline(
    optimal_k,
    color='red',
    linestyle='--',
    label=f'Best K = {optimal_k}'
)

plt.scatter(
    optimal_k,
    max(silhouette_scores),
    color='red',
    s=80
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score vs K")
plt.xticks(K)
plt.grid(True, alpha=0.3)
plt.legend()

plt.show()

```

```python
# Select the K with the highest Silhouette Score

optimal_k = K[silhouette_scores.index(max(silhouette_scores))]

print("Optimal K:", optimal_k)
print("Highest Silhouette Score:", max(silhouette_scores))
```

```python
# Final K-Means Model

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_scaled)

# Add cluster labels to dataframe
df.loc[X.index, 'KMeans_Cluster'] = cluster_labels
df['KMeans_Cluster'] = df['KMeans_Cluster'].astype('Int64')

print("K-Means Cluster Counts:")
print(df['KMeans_Cluster'].value_counts().sort_index())
```

```python
# 3D visualization of K-Means clusters

fig = plt.figure(figsize=(10, 8))

ax = fig.add_subplot(111, projection='3d')

scatter = ax.scatter(
    X['age'],
    X['income'],
    X['visits'],
    c=cluster_labels,
    cmap='viridis',
    s=50,
    alpha=0.7
)

ax.set_xlabel('Age')
ax.set_ylabel('Income')
ax.set_zlabel('Visits')

ax.set_title('K-Means Clustering: Age, Income and Visits')

plt.show()
```

```python
# K-Means Cluster Profile

cluster_profile = df.groupby('KMeans_Cluster')[cluster_features].mean()

display(cluster_profile)
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage

# Single linkage
single_linkage = linkage(
    X_scaled,
    method='single'
)

plt.figure(figsize=(14, 7))

dendrogram(
    single_linkage,
    truncate_mode='lastp',
    p=30
)

plt.title("Hierarchical Clustering - Single Linkage")
plt.xlabel("Cluster / Observation")
plt.ylabel("Distance")

plt.show()
```

```python
# Average linkage
average_linkage = linkage(
    X_scaled,
    method='average'
)

plt.figure(figsize=(14, 7))

dendrogram(
    average_linkage,
    truncate_mode='lastp',
    p=30
)

plt.title("Hierarchical Clustering - Average Linkage")
plt.xlabel("Cluster / Observation")
plt.ylabel("Distance")

plt.show()
```

```python
# Complete linkage
complete_linkage = linkage(
    X_scaled,
    method='complete'
)

plt.figure(figsize=(14, 7))

dendrogram(
    complete_linkage,
    truncate_mode='lastp',
    p=30
)

plt.title("Hierarchical Clustering - Complete Linkage")
plt.xlabel("Cluster / Observation")
plt.ylabel("Distance")

plt.show()
```

```python
# Complete linkage
complete_linkage = linkage(
    X_scaled,
    method='ward'
)

plt.figure(figsize=(14, 7))

dendrogram(
    complete_linkage,
    truncate_mode='lastp',
    p=30
)

plt.title("Hierarchical Clustering - ward method")
plt.xlabel("Cluster / Observation")
plt.ylabel("Distance")

plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score

linkage_methods = ['single', 'complete', 'average','ward']

hierarchical_results = []

for method in linkage_methods:

    model = AgglomerativeClustering(
        n_clusters=2,
        linkage=method
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    hierarchical_results.append({
        'Linkage': method,
        'Silhouette_Score': score
    })

hierarchical_results = pd.DataFrame(hierarchical_results)

display(hierarchical_results)
```

Principal Component Analysis (PCA)


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.decomposition import PCA
```

```python
# Select relevant numerical features
features = [
    'visits',
    'exposure',
    'children',
    'age',
    'income',
    'health1',
    'health2'
]

X = df[features].copy()

print("Selected features:", features)
print("Dataset shape:", X.shape)
print("\nMissing values:")
print(X.isnull().sum())
```

```python
# Handle missing values using the median
imputer = SimpleImputer(strategy='median')
X_imputed = imputer.fit_transform(X)

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

print("Standardized data shape:", X_scaled.shape)
print("Mean of standardized features:",
      np.round(X_scaled.mean(axis=0), 2))
print("Standard deviation:",
      np.round(X_scaled.std(axis=0), 2))
```

```python
# Fit PCA on standardized data
pca_full = PCA()
X_pca_full = pca_full.fit_transform(X_scaled)

# Explained variance
explained_variance = pca_full.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

pca_results = pd.DataFrame({
    'Principal Component': [
        f'PC{i+1}' for i in range(len(explained_variance))
    ],
    'Explained Variance (%)': explained_variance * 100,
    'Cumulative Variance (%)': cumulative_variance * 100
})

display(pca_results.round(2))
```

```python
plt.figure(figsize=(9, 5))

components = np.arange(1, len(explained_variance) + 1)

plt.plot(
    components,
    explained_variance * 100,
    marker='o'
)

plt.xticks(components)
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance (%)')
plt.title('Scree Plot')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    components,
    cumulative_variance * 100,
    marker='o'
)

plt.axhline(
    y=90,
    linestyle='--',
    label='90% variance threshold'
)

plt.xticks(components)
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance (%)')
plt.title('Cumulative Explained Variance')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Reduce the data to two principal components
pca_2d = PCA(n_components=2)
X_pca_2d = pca_2d.fit_transform(X_scaled)

pca_df = pd.DataFrame(
    X_pca_2d,
    columns=['PC1', 'PC2'],
    index=X.index
)

print("Variance explained by PC1 and PC2:",
      round(pca_2d.explained_variance_ratio_.sum() * 100, 2),
      "%")

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=pca_df,
    x='PC1',
    y='PC2',
    alpha=0.7
)

plt.title('PCA: First Two Principal Components')
plt.xlabel(
    f"PC1 ({pca_2d.explained_variance_ratio_[0]*100:.2f}% variance)"
)
plt.ylabel(
    f"PC2 ({pca_2d.explained_variance_ratio_[1]*100:.2f}% variance)"
)
plt.grid(True, alpha=0.25)
plt.show()
```

```python
loadings = pd.DataFrame(
    pca_2d.components_.T,
    index=features,
    columns=['PC1', 'PC2']
)

display(loadings.round(3))

plt.figure(figsize=(7, 5))

sns.heatmap(
    loadings,
    annot=True,
    cmap='coolwarm',
    center=0,
    fmt='.2f'
)

plt.title('PCA Feature Loadings')
plt.show()
```
